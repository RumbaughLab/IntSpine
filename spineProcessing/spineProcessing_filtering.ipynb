```python
ref = pd.read_excel(r"Z:\2026 - E19 Project collab spine and dendrites\Spine\60x_dataset\MIP\output_analysis\global_results_selected.xlsx")
ref = ref.dropna(subset=['Branch_Type'])
ref['expected_filename'] = ref.apply(
    lambda row: f"{row['File']}-s{int(row['Seg_ID']):03d}.tif", axis=1
)
```

```python
keep_files = set(ref['expected_filename'])
# 3. Path to the folder containing your files
folder_path = Path(r"Z:\2026 - E19 Project collab spine and dendrites\Spine\60x_dataset\MIP_segmented_select")
```

```python

folder_path = Path(r"Z:\2026 - E19 Project collab spine and dendrites\Spine\60x_dataset\MIP_segmented_select")
# Categorize files
to_keep = []
to_remove = []
for file_path in folder_path.glob("*.tif"):
    if file_path.name in keep_files:
        to_keep.append(file_path)
    else:
        to_remove.append(file_path)
# --- Summary Counts ---
print(f"Total .tif files in folder : {len(to_keep) + len(to_remove)}")
print(f"Files to KEEP              : {len(to_keep)}")
print(f"Files to REMOVE            : {len(to_remove)}")
print(f"Target count in DataFrame  : {len(keep_files)}")
# --- Preview Samples ---
print("\n--- Sample files to KEEP (first 3) ---")
for f in to_keep[:3]:
    print("  KEEP:", f.name)
print("\n--- Sample files to REMOVE (first 5) ---")
for f in to_remove[:5]:
    print("  REMOVE:", f.name)
```

```python
for file_path in folder_path.glob("*.tif"):
    if file_path.name not in keep_files:
        print(f"Removing: {file_path.name}")
        file_path.unlink()
```
